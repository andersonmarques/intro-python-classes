# Variáveis em Python

Variáveis são usadas para armazenar valores que podem ser referenciados e manipulados ao longo do código. 
Em Python, a criação de uma variável é simples e não requer declaração explícita de tipo, pois Python é uma linguagem de tipagem dinâmica.

## Como criar variáveis

Para criar uma variável, basta atribuir um valor a um nome usando o operador de atribuição `=`. Por exemplo:


```python
x = 2 + 3
print(x)

```

## Por que usar variáveis?
* No computador, as informações são armazenadas em endereços de memória que são identificados por códigos, geralmente representados em hexadecimal (como 0x7ffe73a6).
* Cada dígito hexadecimal pode representar valores de 0 a 15 (ou 0 a F em notação hexadecimal).
* As variáveis nos ajudam a lembrar onde "guardamos" um dado na memória, pois são como "apelidos".


```python
# Ao invés de lembrar algo como 0x7ffe73a6
# É melhor lembrar do "apelido" como: nome, idade ou endereco

nome = "Anderson"

print(nome)
print(x)
```

# Tipos de dados
* Inteiros (int): Representam números inteiros, como positivos ou negativos, sem casas decimais.
* Real ou Pontos Flutuantes (float): Representam números com casas decimais.
* Strings - cadeia de caracteres (str): Representam textos, delimitados por aspas simples ou duplas.
* Booleanos (bool): Representam valores lógicos: verdadeiro (True) ou falso (False).
* Números Complexos (complex): Representam números com parte real e imaginária.

```python
idade = 25
temperatura = 36.5
endereco = "Rua X, n 234567"
aprovado = True
numero_complexo = 3 + 4j #numComplexo

# usando a função type() podemos descobir o tipo de dado.
print(type(idade))
print(type(temperatura))
print(type(endereco))
print(type(aprovado))
print(type(numero_complexo))

print(numero_complexo)  # Saída: (3+4j)
print(numero_complexo.real)  # Parte real: 3.0
print(numero_complexo.imag)  # Parte imaginária: 4.0
```

## Exemplo de uso (if):

## Exemplo de uso (if-elif-else)

Um engenheiro sanitarista está analisando a classificação da qualidade do ar em uma cidade com base no Índice de Qualidade do Ar (IQA). O IQA é classificado em faixas, de acordo com os critérios abaixo:

* IQA entre 0 e 50: Boa - Qualidade do ar satisfatória, sem riscos à saúde.
* IQA entre 51 e 100: Moderada - Qualidade aceitável, mas pode afetar pessoas muito sensíveis.
* IQA entre 101 e 150: Ruim - Pessoas do grupo sensível podem sentir efeitos adversos.
* IQA acima de 150: Muito Ruim - Toda a população pode ser afetada.

Crie um programa que receba o valor do IQA e classifique a qualidade do ar, exibindo uma mensagem correspondente à faixa do índice.

```python
iqa = float(input("Digite o valor do IQA: \n"))

if 0 <= iqa <= 50:
    print("IQA: Boa")
elif 51 <= iqa <= 100:
    print("IQA: Moderada")
elif 101 <= iqa <= 150:
    print("IQA: Ruim")
elif iqa > 150:
    print("IQA: Muito ruim")
else:
    print("IQA: Valor inválido")
```

### Exemplo de uso (match-case):

```python
operacao = input("Digite a operação desejada (+, -, *, /): \n")
a, b = map(int, input("Digite dois valores inteiros separados por espaço: \n").split())

match operacao:# /
    case "+":
        print(a + b)
    case "-":
        print(a - b)
    case "*":
        print(a * b)
    case "/":
        print(a / b)
    case _:
        print("Operação inválida.")
```

### Exemplo de uso (match-case):
Um engenheiro agrônomo deseja classificar diferentes tipos de solo com base no pH. A classificação segue os critérios abaixo:
* Menor que 5.5: Solo ácido.
* Entre 5.5 e 6.5: Solo neutro.
* Maior que 6.5: Solo básico.
Crie um programa que receba o valor do pH e informe a classificação do solo.

```python
# Resolução utilizando if-elif-else:
ph = float(input("Digite o valor do pH do solo: \n"))

if ph < 5.5:
    print("Solo ácido!")
elif 5.5 <= ph <= 6.5:
    print("Solo neutro!")
else:
    print("Solo básico!")
```

```python
# Resolução com match-case:
ph = float(input("Digite o valor do pH do solo: \n"))

match ph:
    case _ if ph < 5.5:
        print("Solo ácido!")
    case _ if 5.5 <= ph <= 6.5:
        print("Solo neutro!")
    case _ if ph > 6.5:
        print("Solo básico!")
```

```python
ph = float(input("Digite o valor do pH do solo: \n"))
saida = "ácido" if ph < 5.5 else "neutro" 

print(f"O solo é {saida}!")
```

# Já ouviram falar de somatório? 

* Uma maneira interessante de compreender os laços de repetição é utilizá-los para resolver somatórios. Na matemática, frequentemente lidamos com expressões que envolvem a soma de um grande número de parcelas. 
* Para simplificar essa notação, utilizamos o símbolo de somatório ( Σ ), que é amplamente aplicado em áreas como cálculo integral, estatística, telecomunicações e computação.

### Usando for para um somatório matemático

* O somatório é uma operação matemática representada pelo símbolo  **Σ** (sigma maiúsculo).
* Em Python, podemos calcular um somatório usando um laço `for`.


### Representação Matemática do Somatório

$$
S = \sum_{i=1}^{N} i
$$

Onde:

-  S : representa o resultado do somatório;
- Σ  (sigma) : indica a operação de soma;
-  i = 1 : Representa o limite inferior, será a variável de iteração, que começa em 1 e vai até  N;
-  N : Representa o limite superior, será o valor final fornecido pelo usuário;
- i : Representa uma observação da série.


```python
# Somatório de 1 a N
n = int(input("Digite um número N para calcular o somatório de 1 até N: "))
soma = 0
for i in range(1, n + 1): # O que é a função range()? É uma função que gera uma sequência de números
    soma += i # soma = soma + i
print(f"O somatório de 1 até {n} é: {soma}")
```

### E sobre o produto na matemática, já ouviram falar?

* Assim como o somatório **(Σ)** representa a soma de vários termos, o produto é representado pelo símbolo pi maiúsculo **(Π)** e indica a multiplicação de uma sequência de números. 
* Esse conceito é amplamente utilizado em áreas como análise combinatória, probabilidade, cálculo e computação, sendo essencial para a representação de produtos sucessivos sem a necessidade de escrever todas as multiplicações manualmente.
* Em Python, podemos calcular produtos usando loops `for`.

O **produto** de 1 até N, também conhecido como **fatorial**, é representado matematicamente por:

$$
P = \prod_{i=1}^{N} i = N!
$$

Ou seja, multiplicamos todos os números inteiros de 1 até N.

Onde:

* P : representa o resultado do produto;
* ∏ (pi maiúsculo) : indica a operação de multiplicação sucessiva;
* i = 1 : representa o limite inferior, ou seja, a variável de iteração que começa em 1 e vai até N;
* N : representa o limite superior, sendo o valor final fornecido pelo usuário;
* i : representa uma observação da série, ou seja, cada número da sequência sendo multiplicado.


```python
# Produto de 1 até N (Fatorial)
n = int(input("Digite um número para calcular o fatorial: "))
produto = 1 # É importante inicializar o produto com 1, pois 0! = 1
for i in range(1, n + 1): # ex: n = 5 -> 1 * 2 * 3 * 4 * 5
    produto *= i # produto = produto * i
print(f"O produto de 1 até {n} (fatorial) é: {produto}")

```

## 3. Estrutura while

* O `while` é usado quando queremos repetir um bloco de código enquanto uma condição for verdadeira.



```python
# Contagem regressiva com while
contador = int(input("Digite um número para iniciar a contagem regressiva: "))
while contador > 0: # enquanto a condição for verdadeira
    print(contador)
    contador -= 1
print("Tempo esgotado!")

```

# 🌀 Hipnose com `continue` e `break` em Python 🌀

🔮 O número 7 desapareceu da sua mente! (continue)

Imagine que você está hipnotizado! Agora, toda vez que contar de 1 a 10, o número 7 simplesmente não existe.

Segue o código mágico:

```python
print("Hipnose iniciada... Agora, ao contar de 1 a 10, o número 7 sumiu! 🌀")

for i in range(1, 11):  
    if i == 7:
        print("🌀-🌀")
        continue  # O número 7 será ignorado!
    print(i)

print("😵 O que aconteceu?! Onde foi parar o 7?!")
```

🔮 Você esqueceu como contar depois do 3! (`break`)

Dessa vez, a hipnose é mais forte! Você começa a contar, mas assim que chega no número 3, simplesmente não consegue mais continuar.

Vamos ver o feitiço em ação:

```python
print("Hipnose nível avançado... Agora, você não consegue contar depois do 3! 😵‍💫")

for i in range(1, 11):  
    if i > 3:
        print("😵‍💫")
        break  # O loop para imediatamente!
    print(i)

print("🤯 O que houve? Para onde foram os números depois do 3?!")
```

## Breve exercício 🚀

Represente os seguintes somatórios em python:
* a) 
$$
\sum_{i=1}^{n} x_{i} y_{i}
$$

* b) 
$$
\sum_{i=1}^{n} (a_{i} - b_{i})
$$

* c) 
$$
\sum_{i=1}^{n} i^i
$$

```python
# a)

# Entrada do número de pares (x, y)
n = int(input("Digite a quantidade de pares (x, y): "))

# Inicializar o somatório
somatorio = 0

# Loop para ler os valores e calcular o somatório
for i in range(1, n + 1):
    x = float(input(f"Digite o valor de x_{i}: "))
    y = float(input(f"Digite o valor de y_{i}: "))
    somatorio += x * y

# Exibir o resultado
print(f"O valor do somatório é: {somatorio}")
```

```python
# Com list comprehension
# Entrada do número de pares (x, y)
n = int(input("Digite a quantidade de pares (x, y): "))

# Cálculo do somatório usando compreensão de listas
somatorio = sum(float(input(f"Digite o valor de x_{i}: ")) * float(input(f"Digite o valor de y_{i}: ")) for i in range(1, n + 1))

# Exibir o resultado
print(f"O valor do somatório é: {somatorio}")
```
